```python
# Część 1: Wczytaj dane
df = spark.read.json("cwiczenia/RTA/transactions_10k.jsonl")

print(f"Liczba rekordów: {df.count()}")
df.printSchema()
```

```python
df.show(10, truncate=False)
```

```python
df.toPandas()
```

```python
from pyspark.sql.functions import to_timestamp, col

df = df.withColumn("timestamp", to_timestamp(col("timestamp"), "yyyy-MM-dd HH:mm:ss")) ## zamiana stringa na timestamp

df.printSchema()  # timestamp powinien być teraz 'timestamp (nullable = true)'
```

```python
# Zadanie 2.1 — Liczba transakcji i suma przychodów per sklep
from pyspark.sql.functions import count, sum as _sum, avg, round as _round

store_summary = (
    df.groupBy("store")
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
        _round(avg("amount"), 2).alias("srednia_PLN"),
    )
    .orderBy("store")
)
store_summary.show()
```

```python
# Zadanie 2.2 - Statystyki per kategoria
# Policz sumę, minimum i maksimum kwoty dla każdej kategorii.
from pyspark.sql.functions import min as _min, max as _max

category_summary = (
    df.groupBy("category")
    .agg(
        _round(_sum("amount"), 2).alias("suma_PLN"),
        _round(_min("amount"),2).alias("min_PLN"),
        _round(_max("amount"),2).alias("max_PLN"),
    )
    .orderBy("category")
)
category_summary.show()
```

```python
# Zadanie 3.1 — Liczba transakcji per godzina (tumbling 1h)
from pyspark.sql.functions import window

hourly = (
    df.groupBy(window("timestamp", "1 hour"))    # okno 1-godzinne
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
    )
    .orderBy("window")
)
hourly.show(truncate=False)
```

```python
hourly.printSchema()
```

```python
(
    hourly
    .select(
        col("window.start").alias("od"),
        col("window.end").alias("do"),
        "liczba_tx",
        "suma_PLN",
    )
    .show(truncate=False)
)
```

```python
# Zadanie 3.2 Okna 30-minutowe per sklep
# Policz transakcje i sumę per sklep w każdym 30-minutowym oknie. Posortuj po oknie, a w ramach okna po sklepie.

from pyspark.sql.functions import window, count, sum as _sum, round as _round

half_hourly = (
    df.groupBy(
        window("timestamp", "30 minutes"),  # okno 30-minutowe
        "store"                             # grupowanie po sklepie
    )
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
    )
    .orderBy("window", "store")  # kolejność
)

half_hourly.show(truncate=False)
```

```python
# Zadanie 3.3 — W której godzinie sklep “Kraków” miał najwyższy przychód? 
# Filtruj najpierw po sklepie, potem zrób okno godzinne, posortuj malejąco po sumie.

from pyspark.sql.functions import window, sum as _sum, round as _round, desc, to_timestamp

krakow_hourly = (
    df.filter(df.store == "Kraków")              # filtr
    .groupBy(window("timestamp", "1 hour"))      # okno 
    .agg(
        _round(_sum("amount"), 2).alias("suma_PLN")
    )
    .orderBy(desc("suma_PLN"))                   # sortowanie malejąco
)

krakow_hourly.show(1, truncate=False)            # top 1
```

```python
# Zadanie 4.1 — Okno 1h, krok 30 minut

sliding = (
    df.groupBy(window("timestamp", "1 hour", "30 minutes"))  # szerokość 1h, krok 30min
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
    )
    .select(
        col("window.start").alias("od"),
        col("window.end").alias("do"),
        "liczba_tx",
        "suma_PLN",
    )
    .orderBy("od")
)
sliding.show(truncate=False)
```

```python
tumbling_rows = (
    df.groupBy(window("timestamp", "1 hour"))
    .agg(count("tx_id"))
    .count()
)
sliding_rows = (
    df.groupBy(window("timestamp", "1 hour", "30 minutes"))
    .agg(count("tx_id"))
    .count()
)
print(f"Tumbling (1h):          {tumbling_rows} okien")
print(f"Sliding  (1h / 30min):  {sliding_rows} okien")

# Odpowiedz w komentarzu: dlaczego sliding ma więcej wierszy?
# TWOJA ODPOWIEDŹ: Skoro okna się nakładają to w tym samym przeddziale czasowym będzie ich więcej niż okien, które się nie nakładają. 
```
