```python
%%file app.py
from flask import Flask, request  # Dodano import request

app = Flask(__name__)

@app.route("/")
def home():
    return "Witaj w systemie monitoringu transakcji!"

@app.route("/hello")
def abc():
    # Zmieniono [] na () oraz dodano brakujący cudzysłów w return
    name = request.args.get('name', 'nieznajomy')
    return f"hello {name}"

if __name__ == "__main__":
    app.run(port=8999)
```

```python
import subprocess
import time
import requests

# 1. Uruchom serwer
server = subprocess.Popen(["python", "app.py"])

# 2. Poczekaj 2-3 sekundy, aż Flask wystartuje
time.sleep(2)

try:
    # 3. Wyślij zapytanie
    response = requests.get("http://localhost:8999/")
    print(f"Status: {response.status_code}")
    print(f"Treść: {response.text}")
    
    # Testuj też drugą trasę
    res_hello = requests.get("http://localhost:8999/hello?name=Ania")
    print(f"Hello route: {res_hello.text}")

finally:
    # 4. Pamiętaj, żeby zabić proces serwera na końcu!
    server.terminate()
```
